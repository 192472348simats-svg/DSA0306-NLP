```python
grammar = {
    "S": [["NP", "VP"]],
    "NP": [[ "N"]],
    "VP": [["V", "NP"]],
    "N": [["ram"], ["mango"]],
    "V": [["eats"]]
}
sentence = input("Enter a sentence: ").lower().split()
def parse(symbol, index):
    if symbol not in grammar:
        if index < len(sentence) and sentence[index] == symbol:
            return True, index + 1
        else:
            return False, index
    for production in grammar[symbol]:
        current_index = index
        success = True
        for sym in production:
            match, current_index = parse(sym, current_index)
            if not match:
                success = False
                break
        if success:
            return True, current_index
    return False, index
matched, final_index = parse("S", 0)
if matched and final_index == len(sentence):
    print("Sentence is accepted.")
else:
    print("Sentence is rejected.")    
```
