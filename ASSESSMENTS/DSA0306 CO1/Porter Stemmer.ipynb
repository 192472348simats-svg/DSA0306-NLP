# **Vani Stemmer**

```python
class VaniStemmer:
    def __init__(self):
        self.endings = ["ness", "ies", "ing", "ed", "ly"]
        self.vowels = "aeiou"
        self.threshold = 2
    def has_vowel(self, word):
        for ch in word:
            if ch in self.vowels:
                return True
        return False
    def detect_ending(self, word):
        for ending in self.endings:
            if word.endswith(ending):
                return ending
        return None
    def has_duplicate(self, word):
        if len(word) < 2:
            return False
        return word[-1] == word[-2]
    def generate_candidates(self, word, ending):
        candidates = []
        base = word[:-len(ending)]
        candidates.append((base, "basic"))
        if self.has_duplicate(base):
            candidates.append((base[:-1], "duplicate_repair"))
        if ending == "ness" and base.endswith("i"):
            candidates.append((base[:-1] + "y", "mutation_repair"))
        if ending == "ies":
            candidates.append((base + "y", "ies_repair"))
        return candidates
    def score(self, candidate, pattern):
        score = 0
        if self.has_vowel(candidate):
            score += 1
        if len(candidate) >= 3:
            score += 1
        if pattern != "basic":
            score += 2
        if self.has_duplicate(candidate):
            score -= 1
        return score
    def stem(self, word):
        word = word.lower()
        ending = self.detect_ending(word)
        if ending is None:
            return word
        candidates = self.generate_candidates(word, ending)
        best_word = word
        best_score = -1
        for candidate, pattern in candidates:
            current_score = self.score(candidate, pattern)
            if current_score > best_score:
                best_score = current_score
                best_word = candidate
        if best_score < self.threshold:
            return word
        return best_word
stemmer = VaniStemmer()
words = [
    "running",
    "stopped",
    "playing",
    "walking",
    "studies",
    "carries",
    "happiness",
    "laziness",
    "slowly",
    "sing",
    "king"
]
print("VANI STEMMER V1")
for word in words:
    print(f"{word:12} -> {stemmer.stem(word)}")
```
