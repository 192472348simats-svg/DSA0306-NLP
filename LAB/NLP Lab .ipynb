## **1. Regular Expressions in Python**

```python
import re

text = "My email is student@gmail.com and my phone is 9876543210."

# Search for email
email = re.search(r'\w+@\w+\.\w+', text)

# Search for phone number
phone = re.search(r'\b\d{10}\b', text)

print("Email:", email.group() if email else "Not found")
print("Phone:", phone.group() if phone else "Not found")
```

## **2. Finite State Automaton**

```python
def accepts(s):
    state = 'q0'

    for ch in s:
        if state == 'q0':
            state = 'q1' if ch == 'a' else 'q0'
        elif state == 'q1':
            state = 'q2' if ch == 'b' else ('q1' if ch == 'a' else 'q0')
        elif state == 'q2':
            state = 'q1' if ch == 'a' else 'q0'

    return state == 'q2'

for s in ["ab", "aab", "abab", "abc", "ba"]:
    print(s, "->", "Accepted" if accepts(s) else "Rejected")
```

## **4. Finite-State Machine for Morphological Parsing**

```python
def generate_plural(noun):
    if noun.endswith(("s", "x", "z", "ch", "sh")):
        return noun + "es"
    elif noun.endswith("y") and noun[-2].lower() not in "aeiou":
        return noun[:-1] + "ies"
    else:
        return noun + "s"

words = ["cat", "dog", "bus", "box", "church", "dish", "baby", "city", "toy"]

for word in words:
    print(word, "->", generate_plural(word))
```

## **6. Bigram N-Gram Model**

```python
from collections import defaultdict
import random

text = "I like natural language processing and I like machine learning"

words = text.lower().split()
bigrams = defaultdict(list)

for a, b in zip(words, words[1:]):
    bigrams[a].append(b)

word = "i"
result = [word]

for _ in range(10):
    if word not in bigrams:
        break
    word = random.choice(bigrams[word])
    result.append(word)

print(" ".join(result))
```

## **8. Stochastic POS Tagging**

```python
import nltk
from collections import defaultdict, Counter

text = "the cat eats the fish"
words = text.split()

# Simple word-tag training data
data = [
    ("the", "DT"), ("cat", "NN"), ("eats", "VBZ"),
    ("the", "DT"), ("fish", "NN")
]

model = defaultdict(Counter)

for word, tag in data:
    model[word][tag] += 1

for word in words:
    tag = model[word].most_common(1)[0][0]
    print(word, "->", tag)
```

## **9. Rule-Based POS Tagging Using Regular Expressions**

```python
import re

def tag(word):
    if re.match(r'.*ing$', word):
        return "VBG"
    elif re.match(r'.*ly$', word):
        return "RB"
    elif re.match(r'.*ed$', word):
        return "VBD"
    elif re.match(r'.*s$', word):
        return "NNS"
    else:
        return "NN"

words = ["running", "quickly", "played", "books", "computer"]

for word in words:
    print(word, "->", tag(word))
```

## **10. Transformation-Based Tagging**

```python
words = ["I", "like", "dogs", "running"]

# Initial tags
tags = ["NN", "NN", "NN", "NN"]

# Transformation rules
for i, word in enumerate(words):
    if word.lower() in ["i"]:
        tags[i] = "PRP"
    elif word.lower() in ["like"]:
        tags[i] = "VBP"
    elif word.endswith("ing"):
        tags[i] = "VBG"
    elif word.endswith("s"):
        tags[i] = "NNS"

for word, tag in zip(words, tags):
    print(word, "->", tag)
```

## **11. Top-Down Parser for CFG**

```python
grammar = {
    "S": [["NP", "VP"]],
    "NP": [["Det", "N"]],
    "VP": [["V", "NP"]],
    "Det": [["the"]],
    "N": [["cat"], ["dog"]],
    "V": [["sees"]]
}

def parse(symbols, words):
    if not symbols:
        return words == []

    first = symbols[0]

    if first not in grammar:
        return words and words[0] == first and parse(symbols[1:], words[1:])

    for rule in grammar[first]:
        if parse(rule + symbols[1:], words):
            return True

    return False

sentence = "the cat sees the dog".split()
print("Accepted" if parse(["S"], sentence) else "Rejected")
```

## **12. Earley Parser**

```python
grammar = {
    "S": [["NP", "VP"]],
    "NP": [["Det", "N"]],
    "VP": [["V", "NP"]],
    "Det": [["the"]],
    "N": [["cat"], ["dog"]],
    "V": [["sees"]]
}

def earley(sentence):
    chart = [set() for _ in range(len(sentence) + 1)]
    chart[0].add(("S", ("NP", "VP"), 0, 0))

    for i in range(len(sentence) + 1):
        changed = True
        while changed:
            changed = False
            for lhs, rhs, dot, start in list(chart[i]):
                if dot < len(rhs):
                    symbol = rhs[dot]

                    if symbol in grammar:
                        for rule in grammar[symbol]:
                            item = (symbol, tuple(rule), 0, i)
                            if item not in chart[i]:
                                chart[i].add(item)
                                changed = True
                    elif i < len(sentence) and symbol == sentence[i]:
                        chart[i + 1].add((lhs, rhs, dot + 1, start))
                else:
                    for l, r, d, s in list(chart[start]):
                        if d < len(r) and r[d] == lhs:
                            item = (l, r, d + 1, s)
                            if item not in chart[i]:
                                chart[i].add(item)
                                changed = True

    return ("S", ("NP", "VP"), 2, 0) in chart[len(sentence)]

sentence = "the cat sees the dog".split()
print("Accepted" if earley(sentence) else "Rejected")
```

## **13. Generate Parse Tree Using CFG**

```python
import nltk

grammar = nltk.CFG.fromstring("""
S -> NP VP
NP -> Det N
VP -> V NP
Det -> 'the'
N -> 'cat' | 'dog'
V -> 'sees'
""")

parser = nltk.ChartParser(grammar)
sentence = "the cat sees the dog".split()

for tree in parser.parse(sentence):
    print(tree)
    tree.pretty_print()
```

## **14. Sentence Agreement Checking**

```python
def check_agreement(subject, verb):
    singular = {"boy", "girl", "cat", "dog"}
    plural = {"boys", "girls", "cats", "dogs"}

    if subject in singular and verb in {"is", "runs", "plays"}:
        return True
    if subject in plural and verb in {"are", "run", "play"}:
        return True
    return False

sentences = [
    ("boy", "runs"),
    ("boys", "run"),
    ("boy", "run"),
    ("girls", "is")
]

for s, v in sentences:
    print(s, v, "->", "Correct" if check_agreement(s, v) else "Incorrect")
```

## **15. Probabilistic CFG Parsing**

```python
import nltk

grammar = nltk.PCFG.fromstring("""
S -> NP VP [1.0]
NP -> Det N [0.5] | 'John' [0.5]
VP -> V NP [1.0]
Det -> 'the' [1.0]
N -> 'dog' [1.0]
V -> 'sees' [1.0]
""")

parser = nltk.ViterbiParser(grammar)

sentence = "John sees the dog".split()

for tree in parser.parse(sentence):
    print(tree)
```

## **18. Simple FOPC Parser**

```python
import re

def parse_fopc(expr):
    pattern = r'([A-Za-z]+)\(([^)]*)\)'
    match = re.fullmatch(pattern, expr.strip())

    if match:
        predicate = match.group(1)
        arguments = match.group(2).split(',')
        return predicate, arguments

    return None

expressions = ["Likes(John,Mary)", "Student(Ravi)"]

for expr in expressions:
    print(expr, "->", parse_fopc(expr))
```

## **20. Information Retrieval Using TF-IDF**

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

documents = [
    "I love machine learning",
    "Machine learning is useful",
    "I like natural language processing"
]

query = ["machine learning"]

vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(documents + query)

scores = cosine_similarity(X[-1], X[:-1]).flatten()

for i, score in enumerate(scores):
    print("Document", i + 1, "Score:", round(score, 3))
```

## **23. Text Coherence Evaluation**

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

text = [
    "Machine learning is a branch of artificial intelligence.",
    "It allows computers to learn from data.",
    "Football is a popular sport."
]

vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(text)

scores = cosine_similarity(X[:-1], X[1:])

print("Coherence score:", round(scores.mean(), 3))
```

## **24. Dialog Act Recognition**

```python
def dialog_act(sentence):
    s = sentence.lower()

    if s.endswith("?"):
        return "Question"
    elif any(word in s for word in ["hello", "hi"]):
        return "Greeting"
    elif any(word in s for word in ["thank", "thanks"]):
        return "Thanking"
    elif any(word in s for word in ["please"]):
        return "Request"
    else:
        return "Statement"

dialog = [
    "Hello!",
    "What is your name?",
    "Please help me.",
    "Thank you."
]

for sentence in dialog:
    print(sentence, "->", dialog_act(sentence))
```
